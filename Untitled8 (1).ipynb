```python
import pandas as pd
import seaborn as sns

# ডাটা সেট লোড করা
titanic = sns.load_dataset("titanic")

# অপ্রয়োজনীয় কলাম বাদ দেওয়া
titanic = titanic.drop(columns=["deck", "class", "who", "adult_male", "embark_town", "alive", "alone"])

# মিসিং ভ্যালু পূরণ করা
titanic["age"] = titanic["age"].fillna(titanic["age"].median())
titanic["embarked"] = titanic["embarked"].fillna(titanic["embarked"].mode()[0])

# ডাটা এনকোডিং বা কনভার্ট করা
titanic["sex"] = titanic["sex"].map({"male": 0, "female": 1})
titanic["pclass"] = titanic["pclass"].map({3: 0, 2: 0, 1: 1})
titanic["sibsp"] = titanic["sibsp"].map(lambda x: 1 if x > 0 else 0)
titanic["parch"] = titanic["parch"].map(lambda x: 1 if x > 0 else 0)
titanic["embarked"] = titanic["embarked"].map({"S": 0, "C": 1, "Q": 2})

# এজ কলাম জিরো থেকে ওয়ানে নরমালাইজ করা
titanic["age"] = (titanic["age"] - titanic["age"].min()) / (titanic["age"].max() - titanic["age"].min())

# ক্লিন ও এনকোড করা ডাটা পরীক্ষা করা
print(titanic.head())

# ফাইলটি CSV আকারে সেভ করা
titanic.to_csv("cleaned_titanic.csv", index=False)
```
